## 1. What is bias in ML models?
Bias is the error that happens when a model is too simple to understand the actual pattern in the data. A high-bias model makes too many assumptions and may perform poorly even on the training data.





## 2. What is variance?
Variance means how much a model’s result changes when the training data changes. 
A high-variance model learns the training data too closely, including its small details and noise, instead of learning the general pattern.

## 3. How do bias and variance relate to underfitting and overfitting?

High bias → Underfitting: The model is too simple and fails to learn enough from the data.

High variance → Overfitting: The model learns the training data too closely and performs poorly on new data.
A good model tries to find a balance between bias and variance so that it performs well on both training and unseen data.

## 4. How can we reduce overfitting?
Overfitting can be reduced by:

Using more training data.

Choosing a simpler model.

Removing unnecessary features.


Using regularization.

Using cross-validation to check the model properly.

Avoiding a model that is unnecessarily complex.
